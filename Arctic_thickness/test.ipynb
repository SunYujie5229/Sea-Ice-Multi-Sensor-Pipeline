```python
import pandas as pd
import matplotlib.pyplot as plt

def plot_freeboard_comparison(dfs, labels):
    plt.figure(figsize=(12, 6))
    
    # 颜色：蓝、橙、绿、红
    colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728'] 
    
    for i, df in enumerate(dfs):
        # 1. 自动适配列名
        # 如果是 L2 数据 (df4)，使用 radar_freeboard_20_ku
        # 否则使用 radar_freeboard
        if 'radar_freeboard_20_ku' in df.columns:
            y_col = 'radar_freeboard_20_ku'
        elif 'radar_freeboard' in df.columns:
            y_col = 'radar_freeboard'
        else:
            print(f"Warning: {labels[i]} 缺少识别的 freeboard 列，已跳过。")
            continue

        # 2. 按照纬度排序（散点图虽然不强求排序，但排序有助于数据清洗和分析）
        df_sorted = df.sort_values(by='lat')
        
        # 3. 绘制散点图
        plt.scatter(df_sorted['lat'], 
                    df_sorted[y_col], 
                    label=labels[i], 
                    color=colors[i], 
                    alpha=0.5, 
                    s=8,           # 点的大小
                    edgecolors='none') 

    plt.xlabel('Latitude (deg)')
    plt.ylabel('Radar Freeboard (m)')
    plt.title('Freeboard Comparison: Processed vs L2 Data')
    
    # 优化图例，使图例中的点变大清晰可见
    plt.legend(markerscale=2, loc='best')
    plt.grid(True, linestyle='--', alpha=0.5)
    
    # 限制 Y 轴范围（可选）：防止异常值导致图像被压扁
    # plt.ylim(-0.5, 2.0) 
    
    plt.tight_layout()
    plt.show()

# 调用示例
# 注意：确保 df1, df2, df3, df4 已经定义
plot_freeboard_comparison([df1, df2, df3, df4], ['threshold_SIT', 'Gauss', 'threshold_fb', 'L2'])
```

```python
import numpy as np
import pandas as pd

def quantify_difference(ref_df, target_dfs, target_labels):
    results = []
    
    # L2 基准列名
    ref_col = 'radar_freeboard_20_ku'
    
    for df, label in zip(target_dfs, target_labels):
        # 1. 识别当前 df 的 freeboard 列名
        if 'radar_freeboard' in df.columns:
            val_col = 'radar_freeboard'
        elif 'radar_freeboard_20_ku' in df.columns:
            val_col = 'radar_freeboard_20_ku'
        else:
            print(f"Skip {label}: column not found")
            continue
            
        # 2. 提取关键列并重命名，避免 merge 后后缀混乱
        # 只取经纬度和数值列
        ref_sub = ref_df[['lat', 'lon', ref_col]].copy()
        ref_sub.columns = ['lat', 'lon', 'val_ref']
        
        target_sub = df[['lat', 'lon', val_col]].copy()
        target_sub.columns = ['lat', 'lon', 'val_test']
        
        # 3. 合并
        combined = pd.merge(ref_sub, target_sub, on=['lat', 'lon'], how='inner')
        
        if combined.empty:
            # 如果经纬度匹配不上，尝试四舍五入对齐
            ref_sub['lat'] = ref_sub['lat'].round(4)
            ref_sub['lon'] = ref_sub['lon'].round(4)
            target_sub['lat'] = target_sub['lat'].round(4)
            target_sub['lon'] = target_sub['lon'].round(4)
            combined = pd.merge(ref_sub, target_sub, on=['lat', 'lon'], how='inner')
            
            if combined.empty:
                print(f"Warning: {label} has no matching coordinates with L2.")
                continue
            
        # 4. 计算差异 (Test - Ref)
        diff = combined['val_test'] - combined['val_ref']
        
        # 5. 统计指标
        metrics = {
            'Method': label,
            'Count': len(diff),
            'Bias (m)': diff.mean(),
            'STD (m)': diff.std(),
            'RMSE (m)': np.sqrt((diff**2).mean()),
            'Median (m)': diff.median()
        }
        results.append(metrics)
    
    return pd.DataFrame(results)

# --- 重新执行 ---
comparison_table = quantify_difference(df4, [df1, df2, df3], ['threshold_SIT', 'Gauss', 'threshold_fb'])
print(comparison_table)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

def plot_quantified_diff(ref_df, target_dfs, labels):
    # 设置绘图风格
    sns.set_theme(style="whitegrid")
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 12))
    
    colors = ['#1f77b4', '#ff7f0e', '#2ca02c']
    ref_col = 'radar_freeboard_20_ku'
    
    all_diffs = []

    for i, (df, label) in enumerate(zip(target_dfs, labels)):
        val_col = 'radar_freeboard' if 'radar_freeboard' in df.columns else 'radar_freeboard_20_ku'
        
        # 数据对齐 (基于经纬度)
        merged = pd.merge(ref_df[['lat', 'lon', ref_col]], 
                          df[['lat', 'lon', val_col]], 
                          on=['lat', 'lon'], how='inner')
        
        if merged.empty:
            continue
            
        # 计算差值: 你的数据 - L2
        merged['diff'] = merged[val_col] - merged[ref_col]
        all_diffs.append(merged['diff'])

        # 图 1: 散点残差图 (随着纬度的变化)
        ax1.scatter(merged['lat'], merged['diff'], 
                    label=f'{label} (bias: {merged["diff"].mean():.3f}m)', 
                    color=colors[i], alpha=0.4, s=5)
        
        # 图 2: 误差概率密度分布
        sns.kdeplot(merged['diff'], ax=ax2, label=label, color=colors[i], fill=True, alpha=0.2)

    # 修饰图 1
    ax1.axhline(0, color='black', linestyle='--', linewidth=1)
    ax1.set_xlabel('Latitude (deg)')
    ax1.set_ylabel('Difference to L2 (m)')
    ax1.set_title('Residuals: (Method - L2) vs Latitude')
    ax1.legend(markerscale=3)

    # 修饰图 2
    ax2.axvline(0, color='black', linestyle='--', linewidth=1)
    ax2.set_xlabel('Difference (m)')
    ax2.set_ylabel('Density')
    ax2.set_title('Error Distribution (Method - L2)')
    ax2.legend()

    plt.tight_layout()
    plt.show()

# --- 调用 ---
# 假设 df1, df2, df3, df4 已准备好
plot_quantified_diff(df4, [df1, df2, df3], ['threshold_SIT', 'Gauss', 'threshold_fb'])
```
